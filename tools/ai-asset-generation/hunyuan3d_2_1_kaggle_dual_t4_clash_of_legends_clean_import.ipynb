# Hunyuan3D 2.1 on Kaggle, using both T4 GPUs

Redesigns each supplied reference image into a new fantasy-game asset concept, then turns the redesigned image into a textured 3D model (`.glb`) with [Tencent Hunyuan3D 2.1](https://github.com/Tencent-Hunyuan/Hunyuan3D-2.1).

**How the two GPUs are used:** the model has two stages, so each stage gets its own T4.

| GPU | Worker | Does |
|---|---|---|
| `cuda:0` | `shape_worker.py` | removes background, then image to untextured mesh (~3.3B-param model) |
| `cuda:1` | `paint_worker.py` | untextured mesh to PBR texture (multiview diffusion + upscaler + baking) |

The two run as separate processes linked by a small folder queue. While GPU 1 is texturing model *N*, GPU 0 is already
building model *N+1*. Neither stage has to share 16 GB with the other, and with a batch of images, throughput roughly doubles.
With only one GPU, the notebook runs the stages one after the other on that GPU.

### Before you run
1. **Settings → Accelerator → GPU T4 x2**
2. **Settings → Internet → On** (needs a phone-verified Kaggle account)
3. Optional: **Add Input → Upload** a folder of images (PNG with a transparent background works best; other images get their background removed automatically).
4. Recommended: **Add Input → Your Datasets →** your Hunyuan3D weights dataset, so the 17.8 GB of weights don't download every session. Build it once with [`hunyuan3d_2_1_weights_builder.ipynb`](https://github.com/ZzwahwahzZ/Clash-of-Legends/blob/main/tools/ai-asset-generation/hunyuan3d_2_1_weights_builder.ipynb).
5. The new image-redesign stage uses SDXL image-to-image on GPU 1 before the Hunyuan3D workers start. Original references are preserved separately; redesigned images become the 3D inputs.\n6. **Run All.** Setup takes about 10 minutes (package installs and compiling the CUDA rasterizer), plus 5–10 minutes more if it has to download the weights.

Results go to `/kaggle/working/hy3d_out/<image name>/textured_mesh.glb` and get zipped into `/kaggle/working/hy3d_results.zip`.
For long batches, use **Save Version → Save & Run All (Commit)** so it keeps running after you close the tab.

## 1. Check the GPUs and locate the attached weights Dataset
If a complete weights Dataset is attached, the notebook reads from `/kaggle/input` directly when there is not enough `/tmp` space to stage a local copy. If no weights Dataset is attached, step 3 downloads the weights and requires sufficient free disk.

```python
%%writefile /tmp/stage_weights.py
"""Copies the weights dataset to local disk with many parallel reads.

Kaggle streams dataset files over the network on first read (~17 MB/s for one sequential reader), so the models
loaded straight from /kaggle/input in 7+ min each. Reading bounded chunks on a small worker pool runs during the install.
Usage: python stage_weights.py <dataset_dir> <local_dir>
"""
import os, sys, threading, time
from concurrent.futures import ThreadPoolExecutor

src, dst = sys.argv[1], sys.argv[2]
CHUNK = 16 << 20  # bounded in-flight work

files = [os.path.join(d, f) for d, _, fs in os.walk(src) for f in fs]
total = sum(os.path.getsize(p) for p in files)
jobs = []
for p in files:
    out = os.path.join(dst, os.path.relpath(p, src))
    os.makedirs(os.path.dirname(out), exist_ok=True)
    size = os.path.getsize(p)
    with open(out, "wb") as f:
        f.truncate(size)
    jobs += [(p, out, off, min(CHUNK, size - off)) for off in range(0, size, CHUNK)]
jobs.sort(key=lambda j: j[2])  # spread early chunks across all files

copied, lock = 0, threading.Lock()


def copy(job):
    global copied
    p, out, off, n = job
    with open(p, "rb") as fi, open(out, "r+b") as fo:
        fi.seek(off)
        fo.seek(off)
        while n:
            buf = fi.read(min(n, 4 << 20))
            if not buf:
                raise IOError(f"short read: {p}")
            fo.write(buf)
            n -= len(buf)
            with lock:
                copied += len(buf)


def report():
    while copied < total:
        time.sleep(30)
        print(f"staged {copied / 2**30:.1f} / {total / 2**30:.1f} GB "
              f"({copied / 2**20 / (time.time() - start):.0f} MB/s)", flush=True)


start = time.time()
threading.Thread(target=report, daemon=True).start()
with ThreadPoolExecutor(max_workers=3) as pool:
    list(pool.map(copy, jobs))
open(os.path.join(dst, "_STAGED"), "w").close()
print(f"STAGED {total / 2**30:.1f} GB in {time.time() - start:.0f}s", flush=True)
```

## 3. Weights: attached dataset, or download (17.8 GB)
If you attached the weights dataset (built once with `hunyuan3d_2_1_weights_builder.ipynb`), this waits for the

```python
%%writefile /tmp/hy3d/prefetch.py
"""Downloads every weight file (~17.8 GB) into one folder, in parallel.

Usage: python prefetch.py <weights_dir>

Layout (the workers and the Kaggle weights dataset both use it):
  <weights_dir>/tencent/Hunyuan3D-2.1/hunyuan3d-dit-v2-1/       shape model      6.9 GB
  <weights_dir>/tencent/Hunyuan3D-2.1/hunyuan3d-paintpbr-v2-1/  texture model    6.4 GB
  <weights_dir>/facebook/dinov2-giant/                          texture encoder  4.2 GB
  <weights_dir>/RealESRGAN_x4plus.pth                           upscaler
  <weights_dir>/u2net/u2net.onnx                                background removal
"""
import os, shutil, sys, urllib.request
from concurrent.futures import ThreadPoolExecutor

try:
    import hf_transfer  # noqa: F401  (several times faster on Kaggle)
    os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "1"
except ImportError:
    pass
from huggingface_hub import snapshot_download

W = sys.argv[1]
DIT = "tencent/Hunyuan3D-2.1/hunyuan3d-dit-v2-1/model.fp16.ckpt"
PAINT = "tencent/Hunyuan3D-2.1/hunyuan3d-paintpbr-v2-1/unet/diffusion_pytorch_model.bin"
DINO = "facebook/dinov2-giant/model.safetensors"
ESRGAN = "RealESRGAN_x4plus.pth"
ESRGAN_ALT = "RealESRGAN_X4plus.pth"
U2NET = "u2net/u2net.onnx"


def have(rel):
    return os.path.exists(os.path.join(W, rel))


def have_esrgan():
    return have(ESRGAN) or have(ESRGAN_ALT)


def hub(repo_id, patterns):
    # One call per repo: parallel calls sharing a local_dir clobber each other's temp files.
    if patterns:
        snapshot_download(repo_id=repo_id, allow_patterns=patterns, local_dir=os.path.join(W, repo_id), token=os.environ.get("HF_TOKEN"))
    return repo_id


def url(link, rel):
    dest = os.path.join(W, rel)
    if not os.path.exists(dest):
        os.makedirs(os.path.dirname(dest), exist_ok=True)
        urllib.request.urlretrieve(link, dest + ".part")
        os.replace(dest + ".part", dest)
    return rel


# Anything already present (e.g. linked in from a partial dataset) is skipped.
tasks = [
    lambda: hub("tencent/Hunyuan3D-2.1", ([] if have(DIT) else ["hunyuan3d-dit-v2-1/*"])
                + ([] if have(PAINT) else ["hunyuan3d-paintpbr-v2-1/*"])),
    lambda: hub("facebook/dinov2-giant", [] if have(DINO) else ["*.json", "*.safetensors", "*.txt"]),
    lambda: (("RealESRGAN already present: " + (ESRGAN if have(ESRGAN) else ESRGAN_ALT)) if have_esrgan() else url("https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth", ESRGAN)),
    lambda: url("https://github.com/danielgatis/rembg/releases/download/v0.0.0/u2net.onnx", U2NET),
]
with ThreadPoolExecutor(max_workers=2) as pool:
    for done in pool.map(lambda task: task(), tasks):
        print("ready:", done)

for repo_id in ("tencent/Hunyuan3D-2.1", "facebook/dinov2-giant"):  # download bookkeeping only
    shutil.rmtree(os.path.join(W, repo_id, ".cache"), ignore_errors=True)

missing = [rel for rel in (DIT, PAINT, DINO, U2NET) if not have(rel)]
if not have_esrgan():
    missing.append(ESRGAN + " (or " + ESRGAN_ALT + ")")
if missing:
    sys.exit(f"Download incomplete, missing: {missing}")
size = sum(os.path.getsize(os.path.join(d, f)) for d, _, fs in os.walk(W) for f in fs)
print(f"All weights present: {size / 2**30:.1f} GB in {W}")
```

## 4. Worker scripts
Each one runs in its own process, pinned to one GPU with `CUDA_VISIBLE_DEVICES`.

```python
%%writefile /tmp/hy3d/shape_worker.py
"""GPU worker 1: image -> untextured mesh. Each finished mesh is queued for paint_worker.py."""
import json, os, sys, time, traceback, gc
import trimesh

cfg = json.load(open(sys.argv[1]))
os.chdir(cfg["repo"])
sys.path.insert(0, os.path.join(cfg["repo"], "hy3dshape"))

import torch
from PIL import Image
from hy3dshape import FaceReducer
from hy3dshape.pipelines import Hunyuan3DDiTFlowMatchingPipeline
from hy3dshape.rembg import BackgroundRemover


def log(msg):
    print(f"[shape {time.strftime('%H:%M:%S')}] {msg}", flush=True)


def enqueue(job):
    tmp = os.path.join(cfg["queue"], f".{job['name']}.tmp")
    with open(tmp, "w") as f:
        json.dump(job, f)
    os.replace(tmp, os.path.join(cfg["queue"], f"{job['name']}.json"))  # atomic: paint never sees half a file


import faulthandler
faulthandler.dump_traceback_later(300, repeat=True)  # while loading: log the stack every 5 min, so a stall shows where
t = time.time()
pipe = Hunyuan3DDiTFlowMatchingPipeline.from_pretrained(
    "tencent/Hunyuan3D-2.1", subfolder="hunyuan3d-dit-v2-1", use_safetensors=False, device="cuda")
rembg = BackgroundRemover()
face_reducer = FaceReducer()
faulthandler.cancel_dump_traceback_later()
log(f"model loaded on {torch.cuda.get_device_name(0)} in {time.time() - t:.0f}s")
open(os.path.join(cfg["queue"], "_SHAPE_READY"), "w").close()

for item in cfg["images"]:
    name, out = item["name"], os.path.join(cfg["out"], item["name"])
    os.makedirs(out, exist_ok=True)
    try:
        existing_obj = os.path.join(out, "white_mesh.obj")
        existing_input = os.path.join(out, "input.png")
        if cfg["skip_existing"] and os.path.isfile(existing_obj) and os.path.isfile(existing_input) and os.path.getsize(existing_obj) > 1024:
            try:
                cached_mesh = trimesh.load(existing_obj, force="mesh", process=False)
                cache_valid = len(cached_mesh.vertices) > 0 and len(cached_mesh.faces) > 0
            except Exception:
                cache_valid = False
            if cache_valid:
                log(f"{name}: verified existing white mesh, skipping shape generation")
                enqueue({"name": name, "dir": out})
                continue
            log(f"{name}: existing white mesh is incomplete; regenerating safely")

        t = time.time()
        torch.cuda.reset_peak_memory_stats()
        with Image.open(item["path"]) as source_image:
            image = source_image.copy()
        if image.mode != "RGBA" or cfg["force_rembg"]:
            processed_image = rembg(image.convert("RGB"))
            image.close()
            image = processed_image
        image.save(os.path.join(out, "input.png"))

        mesh = pipe(
            image=image,
            num_inference_steps=cfg["steps"],
            guidance_scale=cfg["guidance_scale"],
            generator=torch.Generator().manual_seed(cfg["seed"]),
            octree_resolution=cfg["octree_resolution"],
            num_chunks=cfg["num_chunks"],
            output_type="trimesh",
        )[0]
        mesh = face_reducer(mesh, max_facenum=cfg["max_faces"])
        # Write temporary files first; publish final names only after each export completes.
        tmp_glb = os.path.join(out, ".white_mesh.tmp.glb")
        tmp_obj = os.path.join(out, ".white_mesh.tmp.obj")
        mesh.export(tmp_glb)
        mesh.export(tmp_obj, include_normals=False)
        os.replace(tmp_glb, os.path.join(out, "white_mesh.glb"))
        os.replace(tmp_obj, os.path.join(out, "white_mesh.obj"))
        log(f"{name}: {len(mesh.faces)} faces in {time.time() - t:.0f}s "
            f"(peak VRAM {torch.cuda.max_memory_allocated() / 2**30:.1f} GB)")
        enqueue({"name": name, "dir": out})
    except Exception:
        log(f"{name}: FAILED\n{traceback.format_exc()}")
    finally:
        # This worker runs at module scope; drop large per-asset intermediates.
        _image_ref = globals().pop("image", None)
        if _image_ref is not None:
            try:
                _image_ref.close()
            except Exception:
                pass
        for _name in ("mesh", "cached_mesh"):
            globals().pop(_name, None)
        gc.collect()
        torch.cuda.empty_cache()
        try:
            torch.cuda.ipc_collect()
        except Exception:
            pass

open(os.path.join(cfg["queue"], "_SHAPE_DONE"), "w").close()
log("all shapes done")
```

```python
%%writefile /tmp/hy3d/paint_worker.py
"""GPU worker 2: untextured mesh + reference image -> PBR-textured GLB, fed by shape_worker.py."""
import glob, json, os, sys, time, traceback, struct, gc

cfg = json.load(open(sys.argv[1]))
os.chdir(cfg["repo"])
sys.path[:0] = [cfg["repo"], os.path.join(cfg["repo"], "hy3dpaint")]

import base64
import pygltflib
import pymeshlab
import torch
import textureGenPipeline
from textureGenPipeline import Hunyuan3DPaintPipeline, Hunyuan3DPaintConfig
from convert_utils import create_glb_with_pbr_materials


def log(msg):
    print(f"[paint {time.strftime('%H:%M:%S')}] {msg}", flush=True)


def remesh_to_max_faces(mesh_path, remesh_path):
    """Replaces the repo's remesh_mesh, which cuts every mesh to 40k faces (with open3d, which isn't installed).

    That cap silently threw away the detail a higher max_faces setting asked for; this keeps up to max_faces.
    """
    ms = pymeshlab.MeshSet()
    ms.load_new_mesh(mesh_path)
    if ms.current_mesh().face_number() > cfg["max_faces"]:
        ms.meshing_decimation_quadric_edge_collapse(targetfacenum=cfg["max_faces"])
    ms.save_current_mesh(remesh_path, save_textures=False)


textureGenPipeline.remesh_mesh = remesh_to_max_faces  # the pipeline looks it up as a module global at call time


def embed_textures(path):
    """Moves textures into the GLB binary chunk with the right mimeType.

    create_glb_with_pbr_materials stores them as JSON data URIs all labelled image/png (the base colour is a JPEG).
    Web viewers accept that, but it is invalid GLB: Windows 3D Viewer and stricter importers show no texture.
    """
    gltf = pygltflib.GLTF2().load(path)
    blob = bytearray(gltf.binary_blob() or b"")
    changed = False
    for image in gltf.images:
        if not (image.uri or "").startswith("data:"):
            continue
        data = base64.b64decode(image.uri.split(",", 1)[1])
        blob += b"\x00" * (-len(blob) % 4)  # bufferViews must be 4-byte aligned
        gltf.bufferViews.append(pygltflib.BufferView(buffer=0, byteOffset=len(blob), byteLength=len(data)))
        blob += data
        mime = "image/jpeg" if data[:3] == b"\xff\xd8\xff" else "image/png"
        image.uri, image.bufferView, image.mimeType = None, len(gltf.bufferViews) - 1, mime
        changed = True
    if not changed:
        return
    for material in gltf.materials:
        if material.alphaMode in (None, "OPAQUE"):
            material.alphaCutoff = None  # only meaningful with alphaMode MASK
    blob += b"\x00" * (-len(blob) % 4)
    gltf.buffers[0].byteLength = len(blob)
    gltf.set_binary_blob(bytes(blob))
    gltf.save_binary(path)


def next_job():
    jobs = sorted(glob.glob(os.path.join(cfg["queue"], "*.json")), key=os.path.getmtime)
    return jobs[0] if jobs else None


# The texture model always calls snapshot_download(); serve it from the weights folder instead.
import huggingface_hub
_download = huggingface_hub.snapshot_download
def _local_first(repo_id, *args, **kwargs):
    local = os.path.join(cfg["weights"], repo_id)
    return local if os.path.isdir(local) else _download(repo_id, *args, **kwargs)
huggingface_hub.snapshot_download = _local_first  # multiview_utils looks it up at call time

import faulthandler
faulthandler.dump_traceback_later(300, repeat=True)  # while loading: log the stack every 5 min, so a stall shows where
t = time.time()
conf = Hunyuan3DPaintConfig(cfg["texture_views"], cfg["texture_resolution"])
conf.dino_ckpt_path = os.path.join(cfg["weights"], "facebook/dinov2-giant")
conf.realesrgan_ckpt_path = next(
    os.path.join(cfg["weights"], name)
    for name in ("RealESRGAN_x4plus.pth", "RealESRGAN_X4plus.pth")
    if os.path.isfile(os.path.join(cfg["weights"], name))
)
conf.multiview_cfg_path = "hy3dpaint/cfgs/hunyuan-paint-pbr.yaml"
conf.custom_pipeline = "hy3dpaint/hunyuanpaintpbr"
pipe = Hunyuan3DPaintPipeline(conf)
faulthandler.cancel_dump_traceback_later()
log(f"model loaded on {torch.cuda.get_device_name(0)} in {time.time() - t:.0f}s")

while True:
    job_file = next_job()
    if job_file is None:
        # Shape writes _SHAPE_DONE after its last enqueue, so re-check the queue once it appears.
        if os.path.exists(os.path.join(cfg["queue"], "_SHAPE_DONE")) and next_job() is None:
            break
        time.sleep(2)
        continue

    job = json.load(open(job_file))
    os.remove(job_file)
    name, out = job["name"], job["dir"]
    glb_path = os.path.join(out, "textured_mesh.glb")
    if cfg["skip_existing"] and os.path.isfile(glb_path):
        try:
            with open(glb_path, "rb") as existing_file:
                header = existing_file.read(12)
            existing_valid = (len(header) == 12 and header[:4] == b"glTF"
                              and struct.unpack("<I", header[4:8])[0] == 2
                              and struct.unpack("<I", header[8:12])[0] == os.path.getsize(glb_path))
        except OSError:
            existing_valid = False
        if existing_valid:
            embed_textures(glb_path)
            log(f"{name}: verified textured GLB already exists, skipping")
            gc.collect()
            torch.cuda.empty_cache()
            continue
        log(f"{name}: incomplete textured GLB detected; regenerating to a temporary file")
    try:
        t = time.time()
        torch.cuda.reset_peak_memory_stats()
        obj_path = pipe(
            mesh_path=os.path.join(out, "white_mesh.obj"),
            image_path=os.path.join(out, "input.png"),
            output_mesh_path=os.path.join(out, "textured_mesh.obj"),
            save_glb=False,
        )
        base = obj_path[:-len(".obj")]
        tmp_glb_path = os.path.join(out, ".textured_mesh.tmp.glb")
        create_glb_with_pbr_materials(
            obj_path,
            {"albedo": base + ".jpg", "metallic": base + "_metallic.jpg", "roughness": base + "_roughness.jpg"},
            tmp_glb_path,
        )
        embed_textures(tmp_glb_path)
        with open(tmp_glb_path, "rb") as generated_file:
            generated_header = generated_file.read(12)
        if (len(generated_header) != 12 or generated_header[:4] != b"glTF"
                or struct.unpack("<I", generated_header[4:8])[0] != 2
                or struct.unpack("<I", generated_header[8:12])[0] != os.path.getsize(tmp_glb_path)):
            raise RuntimeError("Generated GLB failed header/length validation; final output was not replaced.")
        os.replace(tmp_glb_path, glb_path)
        log(f"{name}: textured in {time.time() - t:.0f}s "
            f"(peak VRAM {torch.cuda.max_memory_allocated() / 2**30:.1f} GB)")
    except torch.cuda.OutOfMemoryError:
        log(f"{name}: OUT OF GPU MEMORY - lower texture_views / texture_resolution in the settings cell")
    except Exception:
        log(f"{name}: FAILED\n{traceback.format_exc()}")
    finally:
        for _name in ("obj_path", "base", "tmp_glb_path", "generated_header"):
            globals().pop(_name, None)
        gc.collect()
        torch.cuda.empty_cache()
        try:
            torch.cuda.ipc_collect()
        except Exception:
            pass

log("all textures done")
```

```python
import glob, os, re, shutil, zipfile, hashlib
from pathlib import Path
from PIL import Image

REPO = "/tmp/Hunyuan3D-2.1"
OUT_DIR = "/kaggle/working/hy3d_out"
IMAGE_EXTRACT_DIR = Path("/kaggle/working/conquer_original_reference_images")
IMAGE_EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

SETTINGS = dict(
    # shape (GPU 0)
    steps=30,
    guidance_scale=5.0,
    octree_resolution=256,
    num_chunks=8000,
    seed=1234,
    max_faces=40000,
    force_rembg=False,
    # texture (GPU 1); conservative settings for a 16 GB T4.
    texture_views=6,
    texture_resolution=512,
    skip_existing=True,
)
MAX_IMAGES = 50
IMAGE_EXTS = (".png", ".jpg", ".jpeg", ".webp", ".bmp")

# Locate the attached Hunyuan weights Dataset so its files are never treated as reference images.
weights_root = Path(DATASET).resolve() if DATASET else None

def is_inside(path, root):
    try:
        Path(path).resolve().relative_to(Path(root).resolve())
        return True
    except (ValueError, OSError):
        return False

# Extract attached reference-image ZIP archives safely. The weights Dataset is excluded.
zip_files = []
for z in Path("/kaggle/input").rglob("*.zip"):
    if weights_root and is_inside(z, weights_root):
        continue
    zip_files.append(z)

# Check the total uncompressed ZIP size before extracting anything.
zip_uncompressed_bytes = 0
for archive in zip_files:
    try:
        with zipfile.ZipFile(archive) as zf:
            zip_uncompressed_bytes += sum(info.file_size for info in zf.infolist() if not info.is_dir())
    except zipfile.BadZipFile:
        pass
working_free_bytes = shutil.disk_usage(IMAGE_EXTRACT_DIR).free
ZIP_DISK_RESERVE_BYTES = 1 << 30
if zip_uncompressed_bytes + ZIP_DISK_RESERVE_BYTES > working_free_bytes:
    raise RuntimeError(
        f"Safe stop before ZIP extraction: need about {zip_uncompressed_bytes / 2**30:.1f} GB "
        f"plus 1 GB reserve, but only {working_free_bytes / 2**30:.1f} GB is free."
    )

for archive in sorted(zip_files):
    try:
        with zipfile.ZipFile(archive) as zf:
            for member in zf.infolist():
                if not member.filename or member.is_dir():
                    continue
                target = (IMAGE_EXTRACT_DIR / member.filename).resolve()
                if not is_inside(target, IMAGE_EXTRACT_DIR):
                    raise RuntimeError(f"Unsafe path in ZIP archive: {member.filename}")
            zf.extractall(IMAGE_EXTRACT_DIR)
        print(f"Extracted reference archive: {archive.name}")
    except zipfile.BadZipFile:
        print(f"Skipping invalid ZIP: {archive}")

# Collect supported images from attached Kaggle inputs and the extracted reference archive.
candidate_roots = [Path("/kaggle/input"), IMAGE_EXTRACT_DIR]
paths = []
for root in candidate_roots:
    if not root.exists():
        continue
    for p in root.rglob("*"):
        if not p.is_file() or p.suffix.lower() not in IMAGE_EXTS:
            continue
        if weights_root and is_inside(p, weights_root):
            continue
        paths.append(str(p))

# Verify image contents and remove exact duplicates by SHA-256.
images, seen_hashes, seen_names = [], set(), set()
invalid_images = []
def sha256_file_stream(path, chunk_bytes=4 << 20):
    digest = hashlib.sha256()
    with open(path, "rb") as stream:
        for block in iter(lambda: stream.read(chunk_bytes), b""):
            digest.update(block)
    return digest.hexdigest()

for p in sorted(set(paths)):
    try:
        digest = sha256_file_stream(p)
        if digest in seen_hashes:
            continue
        with Image.open(p) as im:
            im.verify()
        seen_hashes.add(digest)
        base = re.sub(r"[^A-Za-z0-9_-]+", "_", Path(p).stem).strip("_") or "reference"
        name, suffix = base, 1
        while name in seen_names:
            suffix += 1
            name = f"{base}_{suffix}"
        seen_names.add(name)
        images.append({"name": name, "path": p})
    except Exception as exc:
        invalid_images.append((p, str(exc)))

if invalid_images:
    print(f"Skipped {len(invalid_images)} unreadable or unsupported image file(s).")
if not images:
    raise RuntimeError(
        "No valid reference images were found. In Kaggle, attach the "
        "'Conquer Original Reference Images' Dataset (ZIP or extracted image files), "
        "then rerun this cell. The Hunyuan3D weights Dataset is detected and excluded."
    )

images = images[:MAX_IMAGES]

# Durable recovery: inspect the private Dataset BEFORE SDXL so completed assets consume no GPU.
REDESIGN_ROOT = Path("/kaggle/working/hy3d_redesign")
REDESIGN_ROOT.mkdir(parents=True, exist_ok=True)
from kaggle_secrets import UserSecretsClient
from huggingface_hub import HfApi, hf_hub_download
_HF_TOKEN_EARLY = UserSecretsClient().get_secret("HF_TOKEN")
if not _HF_TOKEN_EARLY:
    raise RuntimeError("Kaggle Secret HF_TOKEN is missing. No generation has started.")
_DATASET_ID_EARLY = "ZzwahwahzZ/clash-of-legends-assets"
_hf_api_early = HfApi(token=_HF_TOKEN_EARLY)
_hf_api_early.create_repo(repo_id=_DATASET_ID_EARLY, repo_type="dataset", private=True, exist_ok=True)
if not getattr(_hf_api_early.repo_info(repo_id=_DATASET_ID_EARLY, repo_type="dataset"), "private", False):
    raise RuntimeError("Recovery Dataset must be private; stopping before generation.")
_remote_early = set(_hf_api_early.list_repo_files(repo_id=_DATASET_ID_EARLY, repo_type="dataset"))
_remote_complete = set()
for _rp in _remote_early:
    _parts = _rp.split("/")
    if len(_parts) == 4 and _parts[0] == "generated" and _parts[1] == "assets" and _parts[3] == "textured_mesh.glb":
        _remote_complete.add(_parts[2])
_pending_images = []
_restored_redesigns = 0
for _item in images:
    _base_name = _item["name"]
    _final_name = f"{_base_name}_redesigned"
    if _final_name in _remote_complete:
        print(f"SKIP before SDXL: {_final_name} already exists in the private Dataset.")
        continue
    _redesign_file = REDESIGN_ROOT / _base_name / "redesigned_input.png"
    _remote_redesign = f"pipeline_state/redesign/{_base_name}/redesigned_input.png"
    if not _redesign_file.is_file() and _remote_redesign in _remote_early:
        _redesign_file.parent.mkdir(parents=True, exist_ok=True)
        _hf_recovered = hf_hub_download(repo_id=_DATASET_ID_EARLY, repo_type="dataset",
            filename=_remote_redesign, token=_HF_TOKEN_EARLY, local_dir="/kaggle/working/hf_recovery")
        shutil.copy2(_hf_recovered, _redesign_file)
        _restored_redesigns += 1
    _item["recovery_redesign_path"] = str(_redesign_file) if _redesign_file.is_file() else None
    _pending_images.append(_item)
images = _pending_images
print(f"Recovery scan: skipped {len(_remote_complete)} remotely completed assets before SDXL; restored {_restored_redesigns} saved redesign(s).")
print(f"Ready for this session: {len(images)} unique reference image(s).")
for item in images:
    print(f"  {item['name']}: {item['path']}")
```

## 5A. AI redesign of reference images (before 3D generation)
This stage preserves each original input, creates a newly designed image with SDXL image-to-image, saves a side-by-side comparison and metrics, then changes each image path so the existing Hunyuan3D workers consume the redesigned image. The similarity score is a visual diagnostic, not a legal clearance test. GPU 1 is released before the Hunyuan3D workers start.

```python
# Stage 5A: create genuinely redesigned image inputs before Hunyuan3D.
# The source images remain untouched. Hunyuan3D receives only the newly generated images.
from pathlib import Path
from PIL import Image, ImageOps, ImageDraw
import json, os, re, gc, time
import numpy as np
import torch
from diffusers import AutoPipelineForImage2Image
from skimage.metrics import structural_similarity

if not images:
    print("All supplied assets are already complete remotely; skipping SDXL and preserving GPU time.")
else:
    REDESIGN_MODEL = "stabilityai/stable-diffusion-xl-base-1.0"
    REDESIGN_ROOT = Path("/kaggle/working/hy3d_redesign")
    REDESIGN_ROOT.mkdir(parents=True, exist_ok=True)
    REDESIGN_STRENGTH = 0.64
    REDESIGN_STEPS = 28
    REDESIGN_GUIDANCE = 6.5
    REDESIGN_SEED = 73184
    REDESIGN_MAX_SIDE = 1024
    REDESIGN_RETRY_SIMILARITY = 0.93
    REDESIGN_PROMPTS = {}  # Optional: {"asset_file_name": "your asset-specific redesign prompt"}
    REDESIGN_SKIP_EXISTING = True

    if torch.cuda.device_count() < 2:
        raise RuntimeError("The existing dual-T4 notebook requires two GPUs; image redesign is assigned to GPU 1.")

    import psutil, subprocess
    ram_before_sdxl_gb = psutil.virtual_memory().available / (1024 ** 3)
    gpu_before_sdxl = subprocess.run(
        ["nvidia-smi", "--query-gpu=index,memory.free", "--format=csv,noheader,nounits"],
        capture_output=True, text=True, check=True
    ).stdout.strip().splitlines()
    gpu1_free_mb = int(gpu_before_sdxl[1].split(",")[1].strip())
    print(f"SDXL preflight: host RAM available={ram_before_sdxl_gb:.1f} GB | GPU 1 free={gpu1_free_mb/1024:.1f} GB")
    if ram_before_sdxl_gb < 12 or gpu1_free_mb < 7000:
        raise RuntimeError("Safe stop before SDXL load: insufficient free host RAM or GPU 1 VRAM. No redesign was started.")

    def prepare_sdxl_image(source_path):
        with Image.open(source_path) as source:
            im = source.copy()
        if im.mode == "RGBA":
            bg = Image.new("RGB", im.size, (128, 128, 128))
            bg.paste(im, mask=im.getchannel("A"))
            im = bg
        else:
            im = im.convert("RGB")
        # Keep the complete object and its aspect ratio; pad rather than crop.
        im = ImageOps.pad(im, (1024, 1024), method=Image.Resampling.LANCZOS, color=(128, 128, 128), centering=(0.5, 0.5))
        return im

    def image_metrics(original, redesigned):
        a = np.asarray(ImageOps.fit(original.convert("RGB"), (512, 512)), dtype=np.uint8)
        b = np.asarray(ImageOps.fit(redesigned.convert("RGB"), (512, 512)), dtype=np.uint8)
        similarity = float(structural_similarity(a, b, channel_axis=2, data_range=255))
        mean_abs_diff = float(np.abs(a.astype(np.float32) - b.astype(np.float32)).mean())
        return {"ssim_similarity": similarity, "mean_absolute_pixel_difference": mean_abs_diff}

    def asset_prompt(name):
        custom = REDESIGN_PROMPTS.get(name)
        if custom:
            return custom
        category = re.sub(r"[_-]+", " ", name).strip()
        return (
            f"Create a genuinely original, production-quality fantasy MMORPG game asset concept for: {category}. "
            "Preserve only the broad gameplay function and subject category. Invent a new silhouette, proportions, "
            "construction, surface geometry, ornaments, materials, textures, and decorative language. "
            "Do not reproduce the exact reference design or its distinctive details. "
            "Classic colorful East Asian fantasy game atmosphere, coherent stylized 3D game-art presentation, "
            "single complete subject, centered, full object visible, three-quarter view, neutral plain studio background, "
            "clean readable shape, detailed craftsmanship, no text, no logo, no watermark."
        )

    print("Loading SDXL image-to-image model on GPU 1. This runs before either Hunyuan3D worker starts.")
    redesign_pipe = AutoPipelineForImage2Image.from_pretrained(
        REDESIGN_MODEL,
        torch_dtype=torch.float16,
        variant="fp16",
        use_safetensors=True,
        add_watermarker=False,
    )
    redesign_pipe.enable_model_cpu_offload(gpu_id=1)
    # Configure VAE memory features on the VAE module itself. Some Diffusers
    # pipeline versions do not expose enable_vae_slicing/enable_vae_tiling.
    redesign_pipe.vae.enable_slicing()
    redesign_pipe.vae.enable_tiling()

    # Attention slicing is version-dependent; use the supported API when present.
    if hasattr(redesign_pipe, "enable_attention_slicing"):
        redesign_pipe.enable_attention_slicing("max")
        attention_slicing_status = "enabled via pipeline"
    elif hasattr(redesign_pipe.unet, "set_attention_slice"):
        redesign_pipe.unet.set_attention_slice("max")
        attention_slicing_status = "enabled via UNet"
    else:
        attention_slicing_status = "not exposed by this Diffusers version"

    print(
        "SDXL memory safeguards enabled: CPU offload, VAE slicing/tiling; "
        f"attention slicing {attention_slicing_status}."
    )

    redesign_reports = []
    for index, item in enumerate(images):
        live_ram_gb = psutil.virtual_memory().available / (1024 ** 3)
        live_gpu = subprocess.run(
            ["nvidia-smi", "--query-gpu=index,memory.free", "--format=csv,noheader,nounits"],
            capture_output=True, text=True
        )
        live_gpu_rows = live_gpu.stdout.strip().splitlines() if live_gpu.returncode == 0 else []
        gpu1_live_free_gb = (int(live_gpu_rows[1].split(",")[1].strip()) / 1024
                             if len(live_gpu_rows) > 1 else 0)
        print(f"Before redesign item {index+1}/{len(images)}: RAM available={live_ram_gb:.1f} GB | GPU 1 free={gpu1_live_free_gb:.2f} GB")
        if live_ram_gb < 9.0 or gpu1_live_free_gb < 0.8:
            raise RuntimeError("Safe stop between redesign items: memory safety floor reached. Completed redesign files are preserved; rerun this stage to resume.")
        name = item["name"]
        original_path = Path(item["path"])
        asset_dir = REDESIGN_ROOT / name
        asset_dir.mkdir(parents=True, exist_ok=True)
        original_copy = asset_dir / "reference_original.png"
        redesigned_path = asset_dir / "redesigned_input.png"
        comparison_path = asset_dir / "original_vs_redesigned.png"

        if not original_path.is_file():
            raise FileNotFoundError(f"Input image is missing: {original_path}")

        with Image.open(original_path) as source_image:
            original = source_image.convert("RGB")
        original.save(original_copy)

        if item.get("recovery_redesign_path") and Path(item["recovery_redesign_path"]).is_file():
            redesigned = Image.open(item["recovery_redesign_path"]).convert("RGB")
            print(f"{name}: restored previously completed redesign from the private Dataset.")
        elif REDESIGN_SKIP_EXISTING and redesigned_path.is_file():
            redesigned = Image.open(redesigned_path).convert("RGB")
            print(f"{name}: reusing completed redesigned image.")
        else:
            init_image = prepare_sdxl_image(original_path)
            prompt = asset_prompt(name)
            result = redesign_pipe(
                prompt=prompt,
                negative_prompt="exact copy, identical design, screenshot, game UI, text, logo, watermark, cropped object, duplicate objects, blurry, low quality",
                image=init_image,
                strength=REDESIGN_STRENGTH,
                guidance_scale=REDESIGN_GUIDANCE,
                num_inference_steps=REDESIGN_STEPS,
                generator=torch.Generator(device="cuda:1").manual_seed(REDESIGN_SEED + index),
            ).images[0].convert("RGB")
            metrics = image_metrics(original, result)
            # If the result remains visually too close, retry once with stronger redesign.
            if metrics["ssim_similarity"] >= REDESIGN_RETRY_SIMILARITY:
                result = redesign_pipe(
                    prompt=prompt,
                    negative_prompt="exact copy, identical design, screenshot, game UI, text, logo, watermark, cropped object, duplicate objects, blurry, low quality",
                    image=init_image,
                    strength=0.76,
                    guidance_scale=7.0,
                    num_inference_steps=32,
                    generator=torch.Generator(device="cuda:1").manual_seed(REDESIGN_SEED + 10000 + index),
                ).images[0].convert("RGB")
                metrics = image_metrics(original, result)
            result.save(redesigned_path)
            try:
                init_image.close()
            except Exception:
                pass
            try:
                result.close()
            except Exception:
                pass
            del init_image, result
            torch.cuda.empty_cache()
            gc.collect()

        with Image.open(redesigned_path) as redesigned_file:
            redesigned = redesigned_file.convert("RGB")
        metrics = image_metrics(original, redesigned)
        preview_original = ImageOps.fit(original, (512, 512))
        preview_redesigned = ImageOps.fit(redesigned, (512, 512))
        preview = Image.new("RGB", (1024, 560), "white")
        preview.paste(preview_original, (0, 32))
        preview.paste(preview_redesigned, (512, 32))
        draw = ImageDraw.Draw(preview)
        draw.text((12, 10), "ORIGINAL REFERENCE", fill="black")
        draw.text((524, 10), "REDESIGNED IMAGE", fill="black")
        preview.save(comparison_path)

        report = {
            "asset": name,
            "source_file": str(original_path),
            "original_copy": str(original_copy),
            "redesigned_image": str(redesigned_path),
            "comparison_preview": str(comparison_path),
            "model": REDESIGN_MODEL,
            "strength": REDESIGN_STRENGTH,
            "metrics": metrics,
            "note": "Similarity metrics are visual diagnostics only; they do not determine copyright status or permission to redistribute."
        }
        redesign_reports.append(report)

        # Persist resumable image-stage artifacts immediately; no previous file is deleted.
        try:
            for _local, _remote in (
                (original_copy, f"pipeline_state/redesign/{name}/reference_original.png"),
                (redesigned_path, f"pipeline_state/redesign/{name}/redesigned_input.png"),
                (comparison_path, f"pipeline_state/redesign/{name}/original_vs_redesigned.png"),
            ):
                _hf_api_early.upload_file(path_or_fileobj=str(_local), path_in_repo=_remote,
                    repo_id=_DATASET_ID_EARLY, repo_type="dataset",
                    commit_message=f"Checkpoint redesign artifact {name}")
            print(f"{name}: redesign checkpoint uploaded to private Dataset.")
        except Exception as _checkpoint_exc:
            print(f"WARNING: redesign checkpoint upload failed for {name}; local files remain: {type(_checkpoint_exc).__name__}: {_checkpoint_exc}")

        item["original_path"] = str(original_copy)
        item["redesigned_path"] = str(redesigned_path)
        item["name"] = f"{name}_redesigned"  # keep new outputs separate from any older meshes made from the untouched reference
        item["path"] = str(redesigned_path)  # existing shape_worker and paint_worker now consume the redesigned image
        print(f"{name}: redesigned | SSIM={metrics['ssim_similarity']:.3f} | mean pixel difference={metrics['mean_absolute_pixel_difference']:.1f}")
        for image_obj in (original, redesigned, preview_original, preview_redesigned, preview):
            try:
                image_obj.close()
            except Exception:
                pass
        del original, redesigned, preview_original, preview_redesigned, preview, draw
        gc.collect()
        torch.cuda.empty_cache()

    report_path = REDESIGN_ROOT / "redesign_report.json"
    report_path.write_text(json.dumps(redesign_reports, indent=2), encoding="utf-8")

    # Release the redesign model before the two Hunyuan3D workers load their own models.
    del redesign_pipe
    gc.collect()
    torch.cuda.empty_cache()
    print(f"Image redesign complete: {len(images)} input(s). Hunyuan3D will now use the redesigned images.")
    print("Originals, redesigned images, side-by-side previews, and the report are in:", REDESIGN_ROOT)
```

## 6. Run a resource-aware, resumable batch
The processing order is unchanged. Safeguards monitor host RAM, both GPUs' VRAM and disk space; per-asset temporary buffers are released, and logs remain under `/kaggle/working/hy3d_logs`. Batch size caps session work, while live guards protect peak memory. It never clears prior output or logs. If resources approach the safety floor, workers are stopped and completed files remain available for the next manually started session.


```python
import json, os, shutil, subprocess, sys, time, gc, struct
from pathlib import Path
import psutil

WORK = "/tmp/hy3d"
QUEUE = f"{WORK}/queue"
_required_setup = ("OUT_DIR", "SETTINGS", "REPO", "WEIGHTS", "OFFLINE", "images")
_missing_setup = [name for name in _required_setup if name not in globals()]
if _missing_setup:
    raise RuntimeError(
        "Section 6 depends on the completed setup cells 1–5A. Missing session values: "
        + ", ".join(_missing_setup)
        + ". Run the notebook cells in order; do not run Section 6 by itself."
    )
OUT_DIR = Path(OUT_DIR)
LOG_DIR = Path("/kaggle/working/hy3d_logs")
LOG_DIR.mkdir(parents=True, exist_ok=True)
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Read actual resources in this Kaggle session; published limits are ceilings, not current free capacity.
def gb(n):
    return n / (1024 ** 3)

vm = psutil.virtual_memory()
ram_available_gb = gb(vm.available)
work_free_gb = gb(shutil.disk_usage("/kaggle/working").free)
tmp_free_gb = gb(shutil.disk_usage("/tmp").free)
gpu_lines = subprocess.run(
    ["nvidia-smi", "--query-gpu=index,name,memory.total,memory.free", "--format=csv,noheader,nounits"],
    capture_output=True, text=True, check=True
).stdout.strip().splitlines()
print(f"Session RAM: total={gb(vm.total):.1f} GB | available={ram_available_gb:.1f} GB")
print(f"Disk free: /kaggle/working={work_free_gb:.1f} GB | /tmp={tmp_free_gb:.1f} GB")
print("GPU status:")
for line in gpu_lines:
    print(" ", line)
gpu_start_free_gb = []
for line in gpu_lines:
    try:
        gpu_start_free_gb.append(int(line.split(",")[3].strip()) / 1024)
    except (IndexError, ValueError):
        pass
if len(gpu_start_free_gb) < 2 or min(gpu_start_free_gb) < 2.0:
    raise RuntimeError("Safe stop: both GPUs must be detected with at least 2 GB free VRAM each before workers start.")

# Refuse to start if the session is already under pressure. No files are removed.
# Conservative operational safety margins; these are not Kaggle's published hard limits.
RAM_START_FLOOR_GB, RAM_STOP_FLOOR_GB = 14.0, 8.0
DISK_START_FLOOR_GB, DISK_STOP_FLOOR_GB = 3.0, 1.5
GPU_STOP_FLOOR_GB = 0.75
if ram_available_gb < RAM_START_FLOOR_GB:
    raise RuntimeError(f"Safe stop: RAM available {ram_available_gb:.1f} GB is below {RAM_START_FLOOR_GB:.1f} GB.")
if work_free_gb < DISK_START_FLOOR_GB or tmp_free_gb < DISK_START_FLOOR_GB:
    raise RuntimeError("Safe stop: less than 3 GB free on a required working disk.")

# Session batch cap limits total work; live RAM/VRAM guards protect per-item peak usage.
if ram_available_gb < 20:
    batch_size = 1
elif ram_available_gb < 25:
    batch_size = 2
else:
    batch_size = 3
print(f"Adaptive session batch cap: {batch_size} asset(s).")

# Load the durable upload record from the private Hugging Face Dataset.
from kaggle_secrets import UserSecretsClient
from huggingface_hub import HfApi
DATASET_ID = "ZzwahwahzZ/clash-of-legends-assets"
hf_token = UserSecretsClient().get_secret("HF_TOKEN")
if not hf_token:
    raise RuntimeError("Kaggle Secret HF_TOKEN is missing. No assets were changed.")
api = HfApi(token=hf_token)
api.create_repo(repo_id=DATASET_ID, repo_type="dataset", private=True, exist_ok=True)
repo_info = api.repo_info(repo_id=DATASET_ID, repo_type="dataset")
if not getattr(repo_info, "private", False):
    raise RuntimeError("The destination Dataset is not Private. No assets were uploaded.")

remote_files = set(api.list_repo_files(repo_id=DATASET_ID, repo_type="dataset"))
remote_asset_names = set()
for remote_path in remote_files:
    parts = remote_path.split("/")
    if len(parts) >= 3 and parts[0] == "generated" and parts[-1] == "textured_mesh.glb":
        remote_asset_names.add(parts[-2])

def valid_glb(path):
    try:
        path = Path(path)
        if not path.is_file() or path.stat().st_size < 1024:
            return False
        with path.open("rb") as f:
            header = f.read(12)
        return (len(header) == 12 and header[:4] == b"glTF"
                and struct.unpack("<I", header[4:8])[0] == 2
                and struct.unpack("<I", header[8:12])[0] == path.stat().st_size)
    except OSError:
        return False

# The image list at this point contains the redesigned inputs from Stage 5A.
# Existing local GLBs are retained for Stage 8; remotely verified assets are never regenerated.
pending = []
local_ready = []
for item in images:
    asset_name = item["name"]
    local_glb = OUT_DIR / asset_name / "textured_mesh.glb"
    if valid_glb(local_glb):
        local_ready.append(asset_name)
        continue
    if asset_name in remote_asset_names:
        continue
    pending.append(item)

print(f"Already uploaded remotely: {len(remote_asset_names)} | valid local GLBs awaiting upload: {len(local_ready)}")
print(f"Assets still requiring generation: {len(pending)}")

# Restore shape-stage checkpoints before launching workers, so completed white meshes are reused.
from huggingface_hub import hf_hub_download
_restored_shape = 0
for _item in pending:
    _asset = _item["name"]
    _asset_dir = OUT_DIR / _asset
    _asset_dir.mkdir(parents=True, exist_ok=True)
    for _filename in ("white_mesh.glb", "input.png"):
        _remote_checkpoint = f"pipeline_state/shape/{_asset}/{_filename}"
        _target = _asset_dir / _filename
        if not _target.is_file() and _remote_checkpoint in remote_files:
            try:
                _downloaded = hf_hub_download(repo_id=DATASET_ID, repo_type="dataset",
                    filename=_remote_checkpoint, token=hf_token,
                    local_dir="/kaggle/working/hf_recovery")
                shutil.copy2(_downloaded, _target)
                _restored_shape += 1
            except Exception as _restore_exc:
                print(f"WARNING: could not restore {_remote_checkpoint}: {type(_restore_exc).__name__}: {_restore_exc}")
print(f"Restored { _restored_shape } durable shape-stage file(s) before worker launch.")
batch_names = [x["name"] for x in pending[:batch_size]]
batch_images = pending[:batch_size]
print("This session's batch:", batch_names if batch_names else "No generation required; proceed to Results and Upload.")

# Rebuild only the ephemeral queue from durable local mesh files. Logs live under /kaggle/working.
# Never delete output meshes, redesign inputs, prior logs, or the Hugging Face completion record.
os.makedirs(WORK, exist_ok=True)
shutil.rmtree(QUEUE, ignore_errors=True)
os.makedirs(QUEUE, exist_ok=True)
run_id = time.strftime("%Y%m%d-%H%M%S")
shape_log = str(LOG_DIR / f"shape_{run_id}.log")
paint_log = str(LOG_DIR / f"paint_{run_id}.log")

def launch(script, gpu, config_path):
    log_file = open(shape_log if script == "shape_worker" else paint_log, "a", buffering=1)
    env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(gpu), PYTHONUNBUFFERED="1",
               HY3DGEN_MODELS=WEIGHTS, U2NET_HOME=f"{WEIGHTS}/u2net")
    if OFFLINE:
        env["HF_HUB_OFFLINE"] = "1"
    proc = subprocess.Popen(
        [sys.executable, f"{WORK}/{script}.py", config_path],
        stdout=log_file, stderr=subprocess.STDOUT, env=env
    )
    return proc, log_file

if batch_images:
    config_path = f"{WORK}/config_{run_id}.json"
    with open(config_path, "w") as f:
        json.dump(dict(SETTINGS, repo=REPO, queue=QUEUE, out=str(OUT_DIR),
                       weights=WEIGHTS, images=batch_images), f)

    n_gpus = len(gpu_lines)
    if n_gpus < 2:
        raise RuntimeError("This notebook requires two T4 GPUs; no worker was started.")
    shape, shape_handle = launch("shape_worker", 0, config_path)
    paint = None
    paint_handle = None
    started = time.time()
    last_stats = 0
    stopped_for_pressure = False

    def pump(path, offset):
        if not os.path.exists(path):
            return offset
        with open(path, errors="replace") as f:
            f.seek(offset)
            chunk = f.read()
            offset = f.tell()
        for line in chunk.replace("\r", "\n").splitlines():
            if line.strip() and not any(s in line for s in ("it/s]", "s/it]", "%|")):
                print(line)
        return offset

    offsets = {shape_log: 0, paint_log: 0}
    while True:
        offsets[shape_log] = pump(shape_log, offsets[shape_log])
        if paint is not None:
            offsets[paint_log] = pump(paint_log, offsets[paint_log])

        if shape.poll() is not None:
            # A durable local white mesh is the recovery point; the queue itself is disposable.
            open(f"{QUEUE}/_SHAPE_DONE", "a").close()
            if shape.returncode != 0 and paint is None:
                print("Shape worker exited before the paint worker could start; inspect the preserved log.")
                break

        if paint is None and os.path.exists(f"{QUEUE}/_SHAPE_READY"):
            paint, paint_handle = launch("paint_worker", 1, config_path)

        if shape.poll() is not None and paint is not None and paint.poll() is not None:
            break

        now = time.time()
        if now - last_stats >= 10:
            last_stats = now
            current_ram = gb(psutil.virtual_memory().available)
            current_work = gb(shutil.disk_usage("/kaggle/working").free)
            current_tmp = gb(shutil.disk_usage("/tmp").free)
            gpu_now = subprocess.run(
                ["nvidia-smi", "--query-gpu=index,memory.free", "--format=csv,noheader,nounits"],
                capture_output=True, text=True
            )
            gpu_free_gb = []
            if gpu_now.returncode == 0:
                for row in gpu_now.stdout.strip().splitlines():
                    try:
                        gpu_free_gb.append(int(row.split(",")[1].strip()) / 1024)
                    except (IndexError, ValueError):
                        pass
            gpu_report = ", ".join(f"GPU{i} free={v:.2f} GB" for i, v in enumerate(gpu_free_gb)) or "GPU VRAM unavailable"
            print(f"--- elapsed={now-started:.0f}s | RAM available={current_ram:.1f} GB | "
                  f"disk free working={current_work:.1f} GB / tmp={current_tmp:.1f} GB | {gpu_report}")
            gpu_pressure = bool(gpu_free_gb) and min(gpu_free_gb) < GPU_STOP_FLOOR_GB
            # Stop before host RAM, disk, or either GPU is exhausted.
            if current_ram < RAM_STOP_FLOOR_GB or current_work < DISK_STOP_FLOOR_GB or current_tmp < DISK_STOP_FLOOR_GB or gpu_pressure:
                print("RESOURCE GUARD: stopping workers before further memory/disk pressure. Existing output is preserved.")
                stopped_for_pressure = True
                for proc in (shape, paint):
                    if proc is not None and proc.poll() is None:
                        proc.terminate()
                for proc in (shape, paint):
                    if proc is not None and proc.poll() is None:
                        try:
                            proc.wait(timeout=20)
                        except subprocess.TimeoutExpired:
                            proc.kill()
                break
        time.sleep(3)

    # Close log handles; never remove these logs.
    shape_handle.close()
    if paint_handle is not None:
        paint_handle.close()
    gc.collect()
    print(f"Batch ended after {(time.time()-started)/60:.1f} min | shape exit={shape.returncode} | "
          f"paint exit={None if paint is None else paint.returncode} | resource_guard={stopped_for_pressure}")
else:
    print("No new 3D generation batch is needed. Existing local results are kept for upload.")

# Durable shape-stage recovery: save white meshes, input images, and worker logs.
try:
    _checkpoint_files = []
    for _asset_dir in OUT_DIR.iterdir():
        if _asset_dir.is_dir():
            for _filename in ("white_mesh.glb", "input.png"):
                _candidate = _asset_dir / _filename
                if _candidate.is_file() and _candidate.stat().st_size > 0:
                    _checkpoint_files.append((_candidate, f"pipeline_state/shape/{_asset_dir.name}/{_filename}"))
    for _log in LOG_DIR.glob("*.log"):
        if _log.is_file() and _log.stat().st_size > 0:
            _checkpoint_files.append((_log, f"pipeline_state/logs/{_log.name}"))
    for _local_path, _remote_path in _checkpoint_files:
        api.upload_file(path_or_fileobj=str(_local_path), path_in_repo=_remote_path,
            repo_id=DATASET_ID, repo_type="dataset",
            commit_message=f"Checkpoint recovery artifact {_local_path.name}")
    print(f"Durable recovery checkpoint: uploaded {len(_checkpoint_files)} intermediate artifact(s) to the private Dataset.")
except Exception as _checkpoint_exc:
    print(f"WARNING: intermediate checkpoint upload failed; local files remain: {type(_checkpoint_exc).__name__}: {_checkpoint_exc}")

del hf_token, UserSecretsClient
```

## Troubleshooting
- **`OUT OF GPU MEMORY` in the paint log**: keep `texture_resolution=512` and lower `texture_views` to 6. Shape-stage OOM: lower `octree_resolution` or `num_chunks`.
- **`nvcc not found`**: the session has no GPU attached. Set the accelerator to GPU T4 x2 and restart.
- **Install fails on a package**: Kaggle updates its image regularly. The error usually names the package; loosen that pin in `install.sh` and rerun the cell.
- **Session restarted**: `/tmp` is wiped, so rerun everything. Outputs in `/kaggle/working` survive while the session is alive, and `skip_existing=True` resumes where it stopped.
- Full worker logs: `/tmp/hy3d/shape_worker.log` and `/tmp/hy3d/paint_worker.log`.

## 8. Upload and verify assets in the private project Dataset
The uploader skips assets already present remotely, verifies each exact remote path after upload, and persists a completion checkpoint in `pipeline_state/completed_assets.json`. A session interrupted during upload can safely continue without duplicating verified assets.

Before running this cell, add a Kaggle Secret named `HF_TOKEN` containing your Hugging Face write token. Never paste the token into notebook code or output.


```python
upload_error = None
try:
    from pathlib import Path
    import json, os, time, struct
    from kaggle_secrets import UserSecretsClient
    from huggingface_hub import HfApi
    
    DATASET_ID = "ZzwahwahzZ/clash-of-legends-assets"
    OUT_DIR = Path("/kaggle/working/hy3d_out")
    STATE_PATH = "pipeline_state/completed_assets.json"
    secrets = UserSecretsClient()
    hf_token = secrets.get_secret("HF_TOKEN")
    if not hf_token:
        raise RuntimeError("Kaggle Secret HF_TOKEN is missing. Add it in Add-ons → Secrets; never paste it into a cell.")
    
    api = HfApi(token=hf_token)
    api.create_repo(repo_id=DATASET_ID, repo_type="dataset", private=True, exist_ok=True)
    dataset_info = api.repo_info(repo_id=DATASET_ID, repo_type="dataset")
    if not getattr(dataset_info, "private", False):
        raise RuntimeError("The Hugging Face Dataset is PUBLIC. No upload performed.")
    
    remote_files = set(api.list_repo_files(repo_id=DATASET_ID, repo_type="dataset"))
    uploaded_names = set()
    for remote_path in remote_files:
        parts = remote_path.split("/")
        if len(parts) >= 3 and parts[0] == "generated" and parts[-1] == "textured_mesh.glb":
            uploaded_names.add(parts[-2])
    
    state = {"dataset": DATASET_ID, "updated_utc": None, "completed_assets": {}}
    state_file = OUT_DIR / "completed_assets_state.json"
    if state_file.is_file():
        try:
            state = json.loads(state_file.read_text(encoding="utf-8"))
        except Exception:
            print("Local state file is unreadable; remote repository listing remains the source of truth.")
    state.setdefault("completed_assets", {})
    
    glbs = sorted(OUT_DIR.glob("*/textured_mesh.glb"))
    validated = []
    for p in glbs:
        if p.stat().st_size < 1024:
            print(f"SKIP tiny/invalid file: {p}")
            continue
        with p.open("rb") as f:
            header = f.read(12)
        if (len(header) != 12 or header[:4] != b"glTF"
                or struct.unpack("<I", header[4:8])[0] != 2
                or struct.unpack("<I", header[8:12])[0] != p.stat().st_size):
            print(f"SKIP incomplete/invalid GLB header or declared length: {p}")
            continue
        validated.append(p)
    
    if not validated:
        print("No validated local GLB files are waiting for upload.")
    else:
        for p in validated:
            asset_name = p.parent.name
            dest = f"generated/assets/{asset_name}/textured_mesh.glb"
            if asset_name in uploaded_names:
                # A matching remote asset path is already present; do not upload a duplicate.
                state["completed_assets"][asset_name] = {"dataset_path": next(
                    x for x in remote_files if x.startswith("generated/") and
                    x.endswith(f"/{asset_name}/textured_mesh.glb")), "bytes": p.stat().st_size}
                print(f"ALREADY VERIFIED remotely: {asset_name}")
                continue
    
            api.upload_file(path_or_fileobj=str(p), path_in_repo=dest, repo_id=DATASET_ID,
                            repo_type="dataset", commit_message=f"Add verified Hunyuan3D asset {asset_name}")
            # Verify the exact destination exists after the upload before marking the asset complete.
            refreshed = set(api.list_repo_files(repo_id=DATASET_ID, repo_type="dataset"))
            if dest not in refreshed:
                raise RuntimeError(f"Upload could not be verified for {asset_name}; local output is preserved.")
            remote_files = refreshed
            uploaded_names.add(asset_name)
            state["completed_assets"][asset_name] = {"dataset_path": dest, "bytes": p.stat().st_size}
            print(f"VERIFIED: {asset_name} -> {dest}")
    
            # Persist progress after each individually verified asset, so an interrupted session resumes safely.
            state["updated_utc"] = time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime())
            state_file.write_text(json.dumps(state, indent=2), encoding="utf-8")
            api.upload_file(path_or_fileobj=str(state_file), path_in_repo=STATE_PATH,
                            repo_id=DATASET_ID, repo_type="dataset",
                            commit_message="Update verified Hunyuan3D pipeline checkpoint")
    
        state["updated_utc"] = time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime())
        state_file.write_text(json.dumps(state, indent=2), encoding="utf-8")
        api.upload_file(path_or_fileobj=str(state_file), path_in_repo=STATE_PATH,
                        repo_id=DATASET_ID, repo_type="dataset",
                        commit_message="Save Hunyuan3D pipeline checkpoint")
        print(f"Checkpoint saved. Verified remote assets: {len(state['completed_assets'])}")
        print(f"Private Dataset: https://huggingface.co/datasets/{DATASET_ID}")
    
    # Persist redesigns, outputs and logs as durable recovery artifacts; no delete operations.
    _recovery_roots = [
        (Path("/kaggle/working/hy3d_redesign"), "pipeline_state/redesign"),
        (OUT_DIR, "pipeline_state/output"),
        (Path("/kaggle/working/hy3d_logs"), "pipeline_state/logs"),
    ]
    _saved_recovery = 0
    for _root, _prefix in _recovery_roots:
        if not _root.exists():
            continue
        for _file in _root.rglob("*"):
            if not _file.is_file() or _file.stat().st_size <= 0:
                continue
            if _file.name.startswith(".") or _file.suffix.lower() in (".tmp", ".obj", ".jpg"):
                continue
            _remote = f"{_prefix}/{_file.relative_to(_root).as_posix()}"
            api.upload_file(path_or_fileobj=str(_file), path_in_repo=_remote,
                repo_id=DATASET_ID, repo_type="dataset",
                commit_message=f"Save recovery checkpoint {_file.name}")
            _saved_recovery += 1
    print(f"Recovery artifacts saved to private Dataset: {_saved_recovery} file(s).")

    del hf_token, secrets
    
except Exception as e:
    upload_error = f"{type(e).__name__}: {e}"
    print("Hugging Face upload/checkpoint step failed; local generated files are preserved.")
    print(upload_error)
```

```python
# FINAL SESSION REPORT — uploads a small JSON summary to this GitHub repository.
# Requires Kaggle Secret GITHUB_TOKEN with Contents: Read and write for this repository.
# This cell does not alter, move, or delete any generated 3D assets.
import os, json, time, shutil, subprocess, traceback
from pathlib import Path
import requests

REPORT_REPO = "ZzwahwahzZ/Clash-of-Legends"
REPORT_BRANCH = "main"
REPORT_DIR = "tools/ai-asset-generation/reports"
OUT = Path("/kaggle/working/hy3d_out")
LOGS = Path("/kaggle/working/hy3d_logs")
REPORT_DIR_LOCAL = Path("/kaggle/working/pipeline_reports")
REPORT_DIR_LOCAL.mkdir(parents=True, exist_ok=True)

def _valid_glb(p):
    try:
        import struct
        with p.open("rb") as f: h = f.read(12)
        return p.stat().st_size >= 1024 and len(h) == 12 and h[:4] == b"glTF" and struct.unpack("<I", h[4:8])[0] == 2 and struct.unpack("<I", h[8:12])[0] == p.stat().st_size
    except Exception:
        return False

def _tail(path, limit=80):
    try: return path.read_text(errors="replace").splitlines()[-limit:]
    except Exception as e: return [f"Could not read log: {type(e).__name__}: {e}"]

def _gpu_status():
    try:
        p = subprocess.run(["nvidia-smi", "--query-gpu=index,name,memory.total,memory.free", "--format=csv"], capture_output=True, text=True, timeout=15)
        return (p.stdout or p.stderr).strip().splitlines()
    except Exception as e: return [f"unavailable: {type(e).__name__}: {e}"]

try:
    import psutil
    vm = psutil.virtual_memory()
    ram = {"total_gb": round(vm.total / 2**30, 2), "available_gb": round(vm.available / 2**30, 2)}
except Exception as e:
    ram = {"error": f"{type(e).__name__}: {e}"}

try:
    disk = {p: {"free_gb": round(shutil.disk_usage(p).free / 2**30, 2), "total_gb": round(shutil.disk_usage(p).total / 2**30, 2)} for p in ["/kaggle/working", "/tmp"]}
except Exception as e:
    disk = {"error": f"{type(e).__name__}: {e}"}

textured = sorted(p for p in OUT.glob("*/textured_mesh.glb") if _valid_glb(p)) if OUT.exists() else []
white = sorted(OUT.glob("*/white_mesh.glb")) if OUT.exists() else []
logs = sorted(LOGS.glob("*.log"), key=lambda p: p.stat().st_mtime) if LOGS.exists() else []
state_path = OUT / "completed_assets_state.json"
try:
    upload_state = json.loads(state_path.read_text(encoding="utf-8")) if state_path.is_file() else {}
except Exception as e:
    upload_state = {"read_error": f"{type(e).__name__}: {e}"}

report = {
    "schema_version": 1,
    "project": "Clash of Legends",
    "repository": REPORT_REPO,
    "reported_at_utc": time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime()),
    "session": {
        "kaggle_working_exists": Path("/kaggle/working").exists(),
        "run_id": globals().get("run_id"),
        "batch_names": globals().get("batch_names", []),
        "resource_guard_stopped_workers": globals().get("stopped_for_pressure"),
        "shape_exit_code": getattr(globals().get("shape"), "returncode", None),
        "paint_exit_code": getattr(globals().get("paint"), "returncode", None),
        "duration_seconds": round(time.time() - globals().get("started", time.time()), 1) if "started" in globals() else None,
    },
    "outputs": {
        "validated_textured_glb_count": len(textured),
        "textured_glb": [{"asset": p.parent.name, "path": str(p), "bytes": p.stat().st_size} for p in textured],
        "white_mesh_glb_count": len(white),
        "white_mesh_names": [p.parent.name for p in white],
    },
    "huggingface_upload_checkpoint": upload_state,
    "huggingface_upload_error": globals().get("upload_error"),
    "resources_at_report_time": {"ram": ram, "disk": disk, "gpu": _gpu_status()},
    "worker_logs": [{"name": p.name, "tail": _tail(p)} for p in logs[-6:]],
}
try:
    ip = get_ipython()
    last_error = getattr(getattr(ip, "last_execution_result", None), "error_in_exec", None)
    if last_error is not None:
        report["last_cell_error"] = {"type": type(last_error).__name__, "message": str(last_error)}
except Exception:
    pass

stamp = time.strftime("%Y%m%d-%H%M%S", time.gmtime())
local_report = REPORT_DIR_LOCAL / f"kaggle-session-{stamp}.json"
local_report.write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
(REPORT_DIR_LOCAL / "latest.json").write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
print("Local session report saved:", local_report)

# The credential is read only from Kaggle Secrets and is never printed or included in the report.
try:
    from kaggle_secrets import UserSecretsClient
    github_token = UserSecretsClient().get_secret("GITHUB_TOKEN")
except Exception:
    github_token = None

if not github_token:
    print("GitHub report not uploaded: add Kaggle Secret GITHUB_TOKEN (repo Contents read/write), then rerun this report cell.")
else:
    headers = {"Authorization": f"Bearer {github_token}", "Accept": "application/vnd.github+json", "X-GitHub-Api-Version": "2022-11-28"}
    def _put_report(path, content, message):
        url = f"https://api.github.com/repos/{REPORT_REPO}/contents/{path}"
        old = requests.get(url, headers=headers, params={"ref": REPORT_BRANCH}, timeout=30)
        payload = {"message": message, "content": __import__("base64").b64encode(content.encode("utf-8")).decode("ascii"), "branch": REPORT_BRANCH}
        if old.status_code == 200:
            payload["sha"] = old.json()["sha"]
        elif old.status_code != 404:
            raise RuntimeError(f"GitHub read failed ({old.status_code}): {old.text[:500]}")
        put = requests.put(url, headers=headers, json=payload, timeout=60)
        if put.status_code not in (200, 201):
            raise RuntimeError(f"GitHub report upload failed ({put.status_code}): {put.text[:700]}")
        return put.json().get("content", {}).get("html_url")

    try:
        latest_url = _put_report(f"{REPORT_DIR}/kaggle-session-latest.json", json.dumps(report, ensure_ascii=False, indent=2), "Update latest Kaggle session report")
        archive_url = _put_report(f"{REPORT_DIR}/kaggle-session-{stamp}.json", json.dumps(report, ensure_ascii=False, indent=2), "Add Kaggle session report")
        print("GitHub report uploaded successfully.")
        print("Latest report:", latest_url)
        print("Archived report:", archive_url)
    except Exception as e:
        print(f"GitHub report upload failed; local report is preserved: {type(e).__name__}: {e}")
    finally:
        github_token = None
```
